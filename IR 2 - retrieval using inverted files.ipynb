```python
# Practical 2: Document Retrieval Using Inverted Files

import nltk
import string

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# Download required NLTK data
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)

# Step 1: Create a text file
text = """This is the first word.
This is the second text, Hello! How are you?
This is the third, this is it now."""

with open("file.txt", "w") as file:
    file.write(text)

# Step 2: Read the file
file = open("file.txt", "r")
read = file.read()
file.seek(0)

array = file.readlines()
file.close()

# Step 3: Count the number of lines
line = len(array)

print("Number of lines in file is:", line)

# Step 4: Remove punctuation and convert to lowercase
clean_text = read.translate(
    str.maketrans("", "", string.punctuation)
)

clean_text = clean_text.lower()

print("\nCleaned Text:")
print(clean_text)

# Step 5: Tokenize the text
tokens = word_tokenize(clean_text)

# Step 6: Remove stop words
stop_words = set(stopwords.words("english"))

tokens_without_sw = [
    word for word in tokens
    if word not in stop_words
]

print("\nTokens after removing stop words:")
print(tokens_without_sw)

# Step 7: Create an inverted index
dict = {}

for i in range(line):

    # Tokenize each line
    words = word_tokenize(array[i].lower())

    # Remove punctuation and stop words
    words = [
        word.translate(str.maketrans("", "", string.punctuation))
        for word in words
    ]

    for item in words:

        if item in tokens_without_sw:

            if item not in dict:
                dict[item] = []

            if i + 1 not in dict[item]:
                dict[item].append(i + 1)

# Step 8: Display the inverted index
print("\nInverted Index:")

for word, lines in dict.items():
    print(word, ":", lines)
```
