```python
# Practical 1: 16-Qubit Random Number Generator

# Step 1: Import required libraries
from qiskit import QuantumRegister, ClassicalRegister
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

# Step 2: Initialize quantum and classical registers
q = QuantumRegister(16, 'q')
c = ClassicalRegister(16, 'c')

# Step 3: Create the quantum circuit
circuit = QuantumCircuit(q, c)

# Step 4: Apply Hadamard gate to all 16 qubits
circuit.h(q)

# Step 5: Measure all qubits
circuit.measure(q, c)

## Display the quantum circuit
# print("\n16-Qubit Quantum Circuit:")
# print(circuit.draw())

# Step 6: Initialize the quantum simulator
simulator = AerSimulator()

# Step 7: Transpile the circuit for the simulator
compiled_circuit = transpile(circuit, simulator)

# Step 8: Execute the circuit (1 shot as in the manual)
job = simulator.run(compiled_circuit, shots=1)

# Step 9: Get the result
result = job.result()
counts = result.get_counts(compiled_circuit)

# Step 10: Display the random number
random_number = list(counts.keys())[0]

print("\nExecuting Quantum Circuit...")
print("Generated 16-bit Random Number:", random_number)
print("Decimal Equivalent:", int(random_number, 2))
```
