```python
# Practical 3: 5-Qubit Quantum Fourier Transform

# Step 1: Import required libraries
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
import numpy as np

# Step 2: Initialize 5 qubits and 5 classical bits
circuit = QuantumCircuit(5, 5)

# Step 3: Encode the input state 10101
circuit.x(4)
circuit.x(2)
circuit.x(0)

circuit.barrier()

# Step 4: Apply 5-Qubit Quantum Fourier Transform

for j in range(5):

    # Apply Hadamard gate
    circuit.h(j)

    # Apply controlled phase rotation gates
    for k in range(j + 1, 5):
        angle = np.pi / (2 ** (k - j))
        circuit.cp(angle, k, j)

# Step 5: Apply SWAP gates to reverse qubit order
for i in range(5 // 2):
    circuit.swap(i, 4 - i)

circuit.barrier()

# Step 6: Measure all qubits
circuit.measure(range(5), range(5))

# Display the quantum circuit
print("\n5-Qubit QFT Circuit:")
print(circuit.draw(output='text', fold=-1))

# Step 7: Initialize the quantum simulator
simulator = AerSimulator()

# Step 8: Transpile the circuit
compiled_circuit = transpile(circuit, simulator)

# Step 9: Execute the circuit with 1000 shots
job = simulator.run(compiled_circuit, shots=1000)

# Step 10: Get measurement results
result = job.result()
counts = result.get_counts()

# Step 11: Display the output
print("\n5-Qubit QFT Output:")
print(counts)
```
